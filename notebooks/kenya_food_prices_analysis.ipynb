```python
retail = food_prices[food_prices["pricetype"] == "Retail"]
beans = retail[retail["commodity"] == "Beans (dry)"]

coverage = beans.pivot_table(index="market", columns="year",
                             values="price", aggfunc="count")
coverage = coverage.fillna(0).astype(int)
coverage["total"] = coverage.sum(axis=1)
coverage.sort_values("total", ascending=False).head(10)
```

```python
kitui = beans[beans["market"] == "Kitui"]
kitui_yearly = kitui.groupby("year")["price"].mean().round(2)
all_yearly = beans.groupby("year")["price"].mean().round(2)

compare = pd.DataFrame({"kitui": kitui_yearly, "all_markets": all_yearly})
print(kitui.groupby("year")["price"].count().to_dict())
print(kitui["priceflag"].value_counts())
compare
```

```python
print(markets["market_id"].is_unique)
print(markets[["admin1", "admin2", "latitude", "longitude"]].isna().sum())

price_ids = set(food_prices["market_id"])
market_ids = set(markets["market_id"])
print(len(price_ids), "markets in prices;",
      len(price_ids - market_ids), "missing from the markets file")
```

```python
hola_fix = markets[markets["market"] == "Hola town"].iloc[0]
mask = food_prices["market"] == "Hola (Tana River)"

food_prices["location_imputed"] = False
for col in ["admin1", "admin2", "latitude", "longitude"]:
    food_prices.loc[mask, col] = hola_fix[col]
food_prices.loc[mask, "location_imputed"] = True

print(food_prices[["admin1", "admin2", "latitude", "longitude"]].isna().sum())
print(food_prices["location_imputed"].sum())
```

```python
clean_dir = Path("../data/clean")
clean_dir.mkdir(exist_ok=True)
food_prices.to_csv(clean_dir / "food_prices_clean.csv", index=False)
print(food_prices.shape)
```
